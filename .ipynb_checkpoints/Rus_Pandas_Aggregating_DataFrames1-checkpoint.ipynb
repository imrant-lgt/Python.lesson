````python
# Объедините результаты для A/B/C в список и разделите на общую сумму продаж, чтобы получить долю продаж по типам.
````

````python
def foo(_type):
    sum_sales = sales[sales.type==_type].weekly_sales.sum()/\
                                sales.weekly_sales.sum()
    return sum_sales
````

Также вы проведете расчеты по данным, сгруппированным по двум переменным, чтобы проверить, отличаются ли продажи в зависимости от типа магазина и того, является ли неделя праздничной.*
*Переменная `sales` доступна, а библиотека `pandas` импортирована как `pd`.*

```markdown
### Сводные таблицы (Pivot tables) по одной переменной
Сводные таблицы — это стандартный способ агрегирования данных в электронных таблицах. В pandas сводные таблицы — это, по сути, еще один способ выполнения групповых вычислений. То есть метод `.pivot_table()` является альтернативой `.groupby()`.

В этом упражнении вы выполните вычисления с использованием `.pivot_table()`, чтобы повторить расчеты, которые вы делали на прошлом уроке с помощью `.groupby()`.

Переменная `sales` доступна, а библиотека `pandas` импортирована как `pd`.
```

```markdown
Получите среднее значение `weekly_sales` по типам магазинов с помощью `.pivot_table()` и сохраните результат в `mean_sales_by_type`.
```

````python
mean_sales_by_type = pd.pivot_table(sales,values='weekly_sales',
                                    index=['type'],aggfunc=np.mean)
````

````python
mean_sales_by_type
````

```markdown
Получите среднее и медианное значения (используя функции NumPy) для `weekly_sales` по типам магазинов с помощью `.pivot_table()` и сохраните результат в `mean_med_sales_by_type`.
```

````python
pd.pivot_table(sales,values='weekly_sales',index=['store','type'],aggfunc=[np.mean,np.median])
````

```markdown
Получите среднее значение `weekly_sales` в разрезе типов магазинов (`type`) и признака праздничных дней (`is_holiday`) с помощью `.pivot_table()` и сохраните результат в `mean_sales_by_type_holiday`.
```

````python
pd.pivot_table(sales,values='weekly_sales',index=['type','is_holiday'],aggfunc=np.mean)
````

```markdown
### Заполнение пропущенных значений и подсчет итогов в сводных таблицах
*Метод `.pivot_table()` имеет несколько полезных аргументов, включая `fill_value` и `margins`.*
*`fill_value` заменяет пропущенные значения конкретным значением (импутация). Тема выбора значения для замены достаточно обширна, но самое простое решение — подставить базовое значение (например, 0).*
*`margins` — это параметр для случаев, когда вы строите сводную таблицу по двум переменным, но также хотите увидеть итоги по каждой из них в отдельности: он добавляет строки и столбцы с общими итогами.*
*В этом упражнении вы попрактикуетесь в использовании этих аргументов, что поможет вам анализировать данные еще эффективнее!*
*Переменная `sales` доступна, а библиотека `pandas` импортирована как `pd`.*
```

```markdown
Выведите среднее значение `weekly_sales` по отделам (`department`) и типам магазинов (`type`), заполняя любые пропущенные значения нулями.
```

````python
pd.pivot_table(sales,values='weekly_sales',fill_value=0,aggfunc=np.mean,
              index=['department','type'])
````

```markdown
Выведите среднее значение `weekly_sales` по отделам (`department`) и типам магазинов (`type`), заполняя любые пропущенные значения нулями и подсчитывая общие итоги по строкам и столбцам.
```

````python
pd.pivot_table(sales,values='weekly_sales',aggfunc=np.mean,
               index=['department','type'],fill_value=0).sum(axis=0)
````
